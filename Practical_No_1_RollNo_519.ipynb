```python
# 3. KDE PLOT
plt.figure(figsize=(7,4))
sns.kdeplot(data=df, x='Total Amount', fill=True)
plt.title("KDE Plot - Total Bill")
plt.xlabel("Total Amount")
plt.show()
```

```python
# 4. BOXPLOT
plt.figure(figsize=(7,4))
sns.boxplot(x=df['Total Amount'])
plt.title("Boxplot - Total Amount")
plt.xlabel("Total Amount")
plt.show()
```

```python
# 7. DISTPLOT
# distplot is removed from new Seaborn,
# so histplot + KDE is used
plt.figure(figsize=(7,4))
sns.histplot(data=df, x='Total Amount', kde=True)
plt.title("Distplot - Total Bill")
plt.xlabel("Total Amount")
plt.show()
```

```python
# 11. REGPLOT
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("tips")

sns.regplot(x="tip", y="total_bill", data=df)

plt.title("Regression Plot")
plt.xlabel("Tip")
plt.ylabel("Total Bill")
plt.show()
```

```python
# 12. PAIRPLOT
sns.pairplot(df)
plt.show()
```

```python
# 13. JOINTPLOT
sns.jointplot(
    data=df,
    x='Total Amount',
    y='Price per Unit',
    kind='scatter'
)
plt.title("Joint Plot - Total Amount vs Price per Unit")
plt.xlabel("Total Amount")
plt.ylabel("Price per Unit")
plt.show()
```

```python
# 14. HEATMAP
plt.figure(figsize=(8,6))

# Select only numerical columns
numeric_df = df.select_dtypes(include='number')

correlation = numeric_df.corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
# 15. LINE PLOT
plt.figure(figsize=(7,4))
sns.lineplot(
    data=df,
    x='Quantity',
    y='Total Amount'
)
plt.title("Line Plot")
plt.xlabel("Quantity")
plt.ylabel("Total Amount")
plt.show()
```

```python
# 16. BAR GRAPH
df['Date'] = pd.to_datetime(df['Date']) # Ensure 'Date' column is datetime
df['day'] = df['Date'].dt.day_name() # Ensure 'day' column exists
plt.figure(figsize=(7,4))
sns.barplot(
    data=df,
    x='day',
    y='Total Amount'
)
plt.title("Bar Graph - Day vs Total Amount")
plt.xlabel("Day")
plt.ylabel("Average Total Amount")
plt.show()
```

```python
# 18. VIOLINPLOT
plt.figure(figsize=(7,4))
sns.violinplot(
    data=df,
    x='day',
    y='Total Amount'
)
plt.title("Violin Plot - Day vs Total Amount")
plt.xlabel("Day")
plt.ylabel("Total Amount")
plt.show()
```

```python
# 19. STRIPPLOT
plt.figure(figsize=(7,4))

sns.stripplot(
    data=df,
    x='day',
    y='total_bill'
)

plt.title("Strip Plot - Day vs Total Bill")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()
```

```python
# 20. SWARMPLOT
plt.figure(figsize=(7,4))

sns.swarmplot(
    data=df,
    x='day',
    y='total_bill'
)

plt.title("Swarm Plot - Day vs Total Bill")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()
```

```python
# 21. STACKED BAR GRAPH

import matplotlib.pyplot as plt
import seaborn as sns

data = df.groupby(['day', 'time']).size().unstack()

data.plot(kind='bar', stacked=True)

plt.title("Stacked Bar Graph")
plt.xlabel("Day")
plt.ylabel("Count")
plt.show()
```

```python
# 22. SIDE-BY-SIDE BAR GRAPH

import matplotlib.pyplot as plt
import seaborn as sns

data = df.groupby(['day', 'time']).size().unstack()

data.plot(kind='bar')

plt.title("Side-by-Side Bar Graph")
plt.xlabel("Day")
plt.ylabel("Count")
plt.show()
```
